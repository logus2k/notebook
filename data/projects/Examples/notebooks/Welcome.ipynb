```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

# 1. Setup the data (using meshgrid for the surface)
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2)) # A nice ripple effect

# 2. Initialize the 3D plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"}, figsize=(10, 7))

# 3. Plot the surface
# 'antialiased=True' and a high 'rcount/ccount' make it look smooth
surf = ax.plot_surface(x, y, z, cmap=cm.magma, 
                       linewidth=0, antialiased=True,
                       rcount=100, ccount=100)

# 4. Polish the aesthetics
ax.set_axis_off()  # Removes the box/grid for a cleaner "floating" look
fig.colorbar(surf, shrink=0.5, aspect=10, pad=0.1) # Optional color guide

plt.title("Matplotlib 3D Ripple Surface", fontsize=15, pad=20)
plt.show()
```
