```python
import plotly.express as px
import pandas as pd
import numpy as np

# 1. Minimalist Data
df = pd.DataFrame({
    "angle": np.arange(0, 360, 15),
    "value": np.random.randint(20, 100, 24)
})

# 2. Eye-catching Polar Chart
fig = px.line_polar(
    df, r="value", theta="angle", 
    line_close=True, 
    template="plotly_dark",
    color_discrete_sequence=["#00FFD1"] # Neon Cyan
)

fig.update_traces(fill='toself') # Fills the shape for a "radar" look

# 3. THE FIX: Save and Open
# This creates a file called 'chart.html' in your folder and opens it
fig.write_html("chart.html", auto_open=True)
```
