# Cell Output Test Notebook

Tests streaming output, progress bars, ANSI colors, clear_output, and display_data.

```python
# Test 1: Simple print (stream)
print('Hello from the kernel!')
print('Line 2')
print('Line 3')
```

```python
# Test 2: Streaming output with sleep
import time
for i in range(5):
    print(f'Step {i+1}/5')
    time.sleep(1)
print('Done!')
```

```python
# Test 3: Carriage return progress bar (like tqdm)
import sys, time
for i in range(20):
    pct = (i+1) * 5
    bar = '=' * (pct // 5) + '>' + '.' * (20 - pct // 5)
    sys.stdout.write(f'\r[{bar}] {pct}%')
    sys.stdout.flush()
    time.sleep(0.3)
print('\nFinished!')
```

```python
# Test 4: ANSI colors
print('\033[1mBold text\033[0m')
print('\033[31mRed text\033[0m')
print('\033[32mGreen text\033[0m')
print('\033[33mYellow text\033[0m')
print('\033[34mBlue text\033[0m')
print('\033[1;35mBold Magenta\033[0m')
```

```python
# Test 5: clear_output with wait=True (like training callbacks)
from IPython.display import clear_output, display
import time

for epoch in range(3):
    clear_output(wait=True)
    print(f'Epoch {epoch+1}/3')
    print(f'loss: {0.5 - epoch * 0.15:.4f}')
    print(f'accuracy: {0.6 + epoch * 0.15:.4f}')
    time.sleep(2)

print('Training complete!')
```
